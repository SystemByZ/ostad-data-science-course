```python
import matplotlib.pyplot as plt
from random import sample

# ডেটা জেনারেশন
data = sample(range(1, 1000), 100)

# গ্রাফ তৈরি
fig, ax = plt.subplots(figsize=(7, 4.5))

# bins=20 (বিন সংখ্যা ২০), facecolor/color (বার এর রঙ), edgecolor (বর্ডারের রঙ)
ax.hist(data, bins=20, color='skyblue', edgecolor='black', alpha=0.8)

ax.set_title('Customized Histogram (Bins = 20)', fontsize=12)
ax.set_xlabel('Value Range')
ax.set_ylabel('Frequency')
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('custom_hist.png')
plt.show()
```

```python
from random import sample
import matplotlib.pyplot as plt

# ১. র‍্যান্ডম ডেটা সেট তৈরি
data = sample(range(1, 1000), 100)

# ২. কাস্টম কালার ও Bins সহ হিস্টোগ্রাম প্লট
plt.hist(
    data, 
    bins=20,             # রেঞ্জকে ২০টি ভাগে ভাগ করবে
    color='purple',     # বারের মূল রঙ (HEX/নাম)
    edgecolor='black',   # বারের সীমানার রঙ
    alpha=0.8            # রঙের সচ্ছতা (0.0 থেকে 1.0)
)

plt.title("Customized Histogram")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# ১. ছবি লোড করা (OpenCV দিয়ে)
img = cv2.imread('images.jpeg')

# OpenCV BGR ফরম্যাটে পড়ে, তাই Matplotlib-এ সঠিকভাবে দেখানোর জন্য RGB-তে কনভার্ট করা
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ২. ছবি ক্রপ করা (Y_start:Y_end, X_start:X_end)
cropped_img = img_rgb[50:200, 100:300]

# ৩. ক্রপ করা ছবি প্রদর্শন করা
plt.imshow(cropped_img)
plt.axis('off')  # Axis বাদ দেওয়া
plt.show()
```
