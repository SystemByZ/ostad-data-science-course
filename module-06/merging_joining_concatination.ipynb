```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5],
    "name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "department": ["HR", "IT", "Finance", "IT", "Marketing"]
})
salaries = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5],
    "salary": [50000, 60000, 55000, 70000, 65000],
    "bonus":[2000, 3000, 2500, 4000, 3500]
})
```

```python
employees
```

```python
print(employees)
print(salaries)
```

```python
employees_with_salaries = pd.merge(employees, salaries, on = "employee_id", how = "inner")
print(employees_with_salaries)
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "outer")
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "left")
```

```python
pd.merge(employees, salaries, on = "employee_id", how = "right")
```

```python
pd.concat([employees, salaries], axis = 0)
```

```python
pd.concat([employees, salaries], axis = 1)
```
