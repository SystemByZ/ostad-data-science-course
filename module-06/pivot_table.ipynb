```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "date" : pd.date_range('2026-01-01',periods=20),
    "product": ['a','b','c','d']*5,
    "region":['east','west','south','north','east','west','south','north','east','west','south','north','east','west','south','north','east','west','south','north'],
    "sales": np.random.randint(100,1000,20),
    "units": np.random.randint(10,100,20),
    "Rep": ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']

})
```

```python
df
```

```python
df['month']=df['date'].dt.month_name()
df['quarter']='Q'+ df['date'].dt.quarter.astype(str)
df
```

```python
pivot_table1 = pd.pivot_table(df,values='sales',index='region',columns='product')
print(pivot_table1)
```

```python
pivot_table2 = pd.pivot_table(df,values='sales',index = 'region',columns = 'product',aggfunc = 'max')
print(pivot_table2)
```

```python
pivot_table3 = pd.pivot_table(df,values=['sales','units'],index='region',columns='product',aggfunc='median')
print(pivot_table3)
```
